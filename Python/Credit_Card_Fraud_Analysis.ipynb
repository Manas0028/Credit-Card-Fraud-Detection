```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../Dataset/creditcard.csv")

print("Dataset loaded successfully!")
print(df.shape)
```

```python
df_clean = df.copy()

print("Cleaned dataset created!")
print(df_clean.shape)
```

```python
df_clean.to_csv(
    "../Dataset/creditcard_cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```

```python
class_summary = df_clean.groupby("Class").agg(
    Total_Transactions=("Class", "count"),
    Average_Amount=("Amount", "mean"),
    Median_Amount=("Amount", "median"),
    Total_Amount=("Amount", "sum")
)

class_summary
```

```python
fraud_percentage = (
    df_clean["Class"].value_counts(normalize=True) * 100
)

fraud_percentage
```

```python
fraud_amount_summary = df_clean[
    df_clean["Class"] == 1
]["Amount"].describe()

fraud_amount_summary
```

```python
top_fraud_transactions = (
    df_clean[df_clean["Class"] == 1]
    .sort_values("Amount", ascending=False)
    [["Time", "Amount", "Class"]]
    .head(10)
)

top_fraud_transactions
```

```python
df_clean["Amount_Range"] = pd.cut(
    df_clean["Amount"],
    bins=[-1, 50, 100, 250, 500, 1000, float("inf")],
    labels=["0-50", "50-100", "100-250", "250-500", "500-1000", "1000+"]
)

amount_fraud_analysis = df_clean.groupby(
    "Amount_Range",
    observed=False
).agg(
    Total_Transactions=("Class", "count"),
    Fraud_Transactions=("Class", "sum")
)

amount_fraud_analysis["Fraud_Rate"] = (
    amount_fraud_analysis["Fraud_Transactions"] /
    amount_fraud_analysis["Total_Transactions"] * 100
)

amount_fraud_analysis
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=amount_fraud_analysis.reset_index(),
    x="Amount_Range",
    y="Fraud_Rate"
)

plt.title("Fraud Rate by Transaction Amount Range")
plt.xlabel("Transaction Amount Range")
plt.ylabel("Fraud Rate (%)")
plt.xticks(rotation=0)
plt.show()
```

```python
df_clean["Hour"] = (df_clean["Time"] / 3600) % 24

df_clean[["Time", "Hour"]].head()
```

```python
fraud_by_hour = df_clean.groupby("Hour")["Class"].agg(
    Total_Transactions="count",
    Fraud_Transactions="sum"
)

fraud_by_hour.head()
```

```python
fraud_by_hour["Fraud_Rate"] = (
    fraud_by_hour["Fraud_Transactions"] /
    fraud_by_hour["Total_Transactions"]
) * 100

fraud_by_hour
```

```python
plt.figure(figsize=(12, 6))

sns.lineplot(
    data=fraud_by_hour,
    x=fraud_by_hour.index,
    y="Fraud_Transactions"
)

plt.title("Fraudulent Transactions by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Number of Fraudulent Transactions")
plt.xticks(range(0, 24))
plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.lineplot(
    data=fraud_by_hour,
    x=fraud_by_hour.index,
    y="Fraud_Rate"
)

plt.title("Fraud Rate by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Fraud Rate (%)")
plt.xticks(range(0, 24))
plt.show()
```

```python
correlation = df_clean.corr(numeric_only=True)

correlation["Class"].sort_values(ascending=False)
```

```python
plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation,
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Matrix of Transaction Features")
plt.show()
```

```python
legitimate = df_clean[df_clean["Class"] == 0]
fraud = df_clean[df_clean["Class"] == 1]

print("Legitimate transactions:", len(legitimate))
print("Fraudulent transactions:", len(fraud))
```

```python
print("Average legitimate transaction:",
      legitimate["Amount"].mean())

print("Average fraudulent transaction:",
      fraud["Amount"].mean())
```

```python
print("Median legitimate transaction:",
      legitimate["Amount"].median())

print("Median fraudulent transaction:",
      fraud["Amount"].median())
```

```python
print("===== CREDIT CARD FRAUD EDA SUMMARY =====")

print("\n1. Dataset Overview")
print("Total transactions:", len(df_clean))

print("\n2. Fraud Distribution")
print("Legitimate transactions:", len(legitimate))
print("Fraudulent transactions:", len(fraud))
print("Fraud rate:",
      round((len(fraud) / len(df_clean)) * 100, 3), "%")

print("\n3. Transaction Amount")
print("Average legitimate amount:",
      round(legitimate["Amount"].mean(), 2))
print("Average fraudulent amount:",
      round(fraud["Amount"].mean(), 2))

print("\nMedian legitimate amount:",
      round(legitimate["Amount"].median(), 2))
print("Median fraudulent amount:",
      round(fraud["Amount"].median(), 2))

print("\n4. Highest Fraud Amount")
print("Maximum fraudulent transaction:",
      round(fraud["Amount"].max(), 2))

print("\n5. Highest Fraud-Rate Amount Range")
print(
    amount_fraud_analysis["Fraud_Rate"].idxmax(),
    "with fraud rate:",
    round(amount_fraud_analysis["Fraud_Rate"].max(), 3), "%"
)

print("\n6. Highest Fraud-Rate Hour")
print(
    fraud_by_hour["Fraud_Rate"].idxmax(),
    "with fraud rate:",
    round(fraud_by_hour["Fraud_Rate"].max(), 3), "%"
)
```
